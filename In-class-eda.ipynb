Using `iris` dataset, make a **scatter chart** where 
- x represents `sepal_length`
- y represents `petal_length`

```python

# The indices of the features that we are plotting
sepallength = 0
petallength = 2

plt.figure(figsize=(5, 4))
plt.scatter(iris.iloc[:, sepallength], iris.iloc[:, petallength], label='iris scatter chart')
plt.xlabel('sepal length (cm)')
plt.ylabel('petal length (cm)')
plt.title('Iris dataset: petal length vs sepal length')
plt.legend(loc='lower right')
plt.show()
```

Using `iris` dataset, make a **scatter chart** where 
- x represents `sepal_length`
- y represents `petal_length`
- add colorbar for `species_id`.

```python
# The indices of the features that we are plotting
sepallength = 0
petallength = 2
cm = plt.cm.get_cmap('RdYlBu')
plt.figure(figsize=(5, 4))
sc = plt.scatter(iris.iloc[:, sepallength], iris.iloc[:, petallength], label='iris scatter chart',vmin=0, vmax=20, s=35, cmap=cm)
plt.colorbar(sc)
plt.xlabel('sepal length (cm)')
plt.ylabel('petal length (cm)')
plt.title('Iris dataset: petal length vs sepal length')
plt.legend(loc='lower right')
plt.show()
```

### Seaborn

Using `iris` dataset: 

- make scatter chart
- make jointplot
- make boxplot
- make violinplot
- make swarmplot
- make pairplot

```python
'''
make scatter chart
'''

#create a figure "fig" with axis "ax1" with 3x2 configuration
fig, ax1 = plt.subplots(2,1, sharex='col', figsize=(22,18), gridspec_kw={'hspace': 0.1, 'wspace': 0.5}) 


# 1st plot
sns.set_style("whitegrid");
sns.scatterplot(data=iris, x="sepal_length", y="petal_length", hue="species", ax=ax1[0], legend='brief') 
# 2nd plot
sns.scatterplot(data=iris, x="petal_length", y="petal_width", hue="species", ax=ax1[1], legend='brief') 
fig.savefig("data/snsfig1.png")
```

```python
'''
make strip chart
'''

ax=sns.stripplot(x='species',y='sepal_length',data=iris,jitter=True,edgecolor='gray')
```

```python
'''
- make jointplot
'''

sns.jointplot(x='sepal_length',y='sepal_width',data=iris,size=10)
```

```python
'''
- make boxplot
'''
fig=plt.gcf()
fig.set_size_inches(10,7)
fig =sns.boxplot(x='species',y='petal_length',data=iris)
```

```python
'''
- make violinplot
'''
sns.violinplot(x='species',y='sepal_length',data=iris,size=6)
```

```python
'''
- make swarmplot
'''
sns.set(style="whitegrid")
fig=plt.gcf()
fig.set_size_inches(10,7)
fig = sns.swarmplot(x="species", y="petal_length", data=iris)
```

```python
'''
- make pairplot
'''
sns.set_style("whitegrid");
sns.pairplot(iris, hue="species", size=3);
plt.show()
```

```python
#OR 
sns.pairplot(data=iris,kind='scatter')
```

Plot the following dataset:

- make line chart
- add a moving average

```python
'''
- make line chart
'''
sns.lineplot(data=iris,x='sepal_length', y='sepal_width', hue='species')
```

```python
'''
- add a moving average
'''

rolling_mean = iris.sepal_length.rolling(window=20).mean()
rolling_mean2 = iris.sepal_length.rolling(window=50).mean()

plt.plot(rolling_mean, iris.sepal_width, label='moving average by 20')
plt.plot(rolling_mean2,iris.sepal_width,  label='moving average by 50')
plt.legend(loc='upper right')
plt.show()
```

```python
'''
Heat Map
'''
plt.figure(figsize=(7,4))
sns.heatmap(iris.corr(),annot=True,cmap='summer')
```
